```python

import pandas as pd
import random
from faker import Faker

fake = Faker()
random.seed(42)
Faker.seed(42)

rows = []

universities = [
    "UIT University",
    "NUST",
    "NED University",
    "FAST",
    "Bahria University",
    ""
]

domains = [
    "Data Science",
    "Web Development",
    "UI/UX Design",
    "Cyber Security",
    "AI",
    "web dev",
    "WEB DEV"
]

statuses = [
    "Under Review",
    "Selected",
    "Rejected",
    "selected",
    "REJECTED",
    ""
]

for i in range(100):
    name = fake.name()
    email = fake.email()

    phone = "03" + str(random.randint(0, 9)) + \
            str(random.randint(10000000, 99999999))

    domain = random.choice(domains)
    university = random.choice(universities)
    status = random.choice(statuses)

    date = random.choice([
        "12/05/2026",
        "2026-05-12",
        "15/06/2026"
    ])

    if i % 10 == 0:
        name = "  " + name + "  "

    if i % 12 == 0:
        email = ""

    if i % 15 == 0:
        phone = ""

    if i % 9 == 0:
        university = ""

    rows.append([
        name, email, phone, domain,
        university, date, status
    ])

df = pd.DataFrame(rows, columns=[
    "Applicant Name",
    "Email",
    "Phone",
    "Domain Applied",
    "University",
    "Application Date",
    "Status"
])

df.to_csv("applicants.csv", index=False)

print("CSV created successfully!")
print("Number of rows and columns:", df.shape)

display(df.head(10))
```

The dataset contains 100 rows and 7 columns. All columns currently have the object data type. The initial exact duplicate check returned 0, meaning no completely identical rows were found.

The initial inspection helps identify missing values, inconsistent text formatting, and data type issues before cleaning. Further checks are needed to identify empty strings and variations in domain and status values.

```python
import pandas as pd

df = df.replace(r'^\s*$', pd.NA, regex=True)

print("Missing values before cleaning:")
print(df.isnull().sum())
```

```python

before_name = len(df)
df = df.dropna(subset=["Applicant Name"])
removed_names = before_name - len(df)

before_email = len(df)
df = df.dropna(subset=["Email"])
removed_emails = before_email - len(df)

df["University"] = df["University"].fillna("Not Specified")

df["Status"] = df["Status"].fillna("Under Review")

print("Rows removed due to missing names:", removed_names)
print("Rows removed due to missing emails:", removed_emails)

print("\nMissing values after Step 4:")
print(df.isnull().sum())

print("\nRemaining dataset shape:", df.shape)
```

The dataset initially contained 9 missing Email values, 7 missing Phone values, 26 missing University values, and 15 missing Status values.

Rows with missing Applicant Name were removed because the applicant could not be reliably identified. Rows with missing Email were also removed because email is important for follow-up.

Missing University values were filled with "Not Specified" because this information is not essential for identifying an application. Missing Status values were filled with "Under Review" because these applications may not have been processed yet.

Phone values were left unchanged at this stage for further consideration.

The number of rows removed and the remaining missing values were checked after applying these changes.

```python
exact_duplicates = df.duplicated().sum()

print("Exact duplicate rows:", exact_duplicates)

df = df.drop_duplicates()

print("Dataset shape after removing exact duplicates:", df.shape)
```

```python
duplicate_emails = df["Email"].duplicated().sum()

print("Duplicate email entries:", duplicate_emails)

display(
    df[df["Email"].duplicated(keep=False)]
    .sort_values("Email")
)
```

I checked the dataset for exact duplicate rows using `df.duplicated().sum()`. The result was 0, so no completely identical rows were found.

I also checked for duplicate email addresses because email can help identify repeated applications from the same applicant. The result was 0 duplicate email entries.

No rows were removed during this step because no exact duplicates or duplicate email addresses were detected. The dataset remains at 91 rows and 7 columns.

Email was chosen as a duplicate-checking field because names may be shared by different applicants, while email addresses are generally more useful for identifying repeated applications.

```python
text_columns = [
    "Applicant Name",
    "Email",
    "Phone",
    "Domain Applied",
    "University",
    "Status"
]

for col in text_columns:
    df[col] = df[col].str.strip()

df["Applicant Name"] = df["Applicant Name"].str.title()

df["Email"] = df["Email"].str.lower()

df["University"] = df["University"].str.title()

display(df.head())
```

```python
print("Domains before standardization:")
print(df["Domain Applied"].unique())

domain_mapping = {
    "data science": "Data Science",
    "web dev": "Web Development",
    "web development": "Web Development",
    "ui/ux design": "UI/UX Design",
    "cyber security": "Cyber Security",
    "ai": "AI"
}

df["Domain Applied"] = df["Domain Applied"].str.lower().str.strip()

df["Domain Applied"] = df["Domain Applied"].map(domain_mapping)

print("\nDomains after standardization:")
print(df["Domain Applied"].unique())

print("\nMissing domain values after mapping:")
print(df["Domain Applied"].isnull().sum())
```

```python
df["Status"] = df["Status"].str.strip().str.title()
print("Unique status values:")
print(df["Status"].unique())
```

I removed leading and trailing whitespace from text columns using `str.strip()`. I standardized applicant names and university text formatting and converted email addresses to lowercase.

I used a mapping dictionary to convert inconsistent domain labels, such as "web dev" and "WEB DEV", into the standard label "Web Development".

I also standardized application statuses using title case so that values such as "selected" and "SELECTED" are represented consistently.

Finally, I checked the unique domain and status values to verify the results and identify any unmapped domain variations.

```python
print("Date type before conversion:")
print(df["Application Date"].dtype)

df["Application Date"] = pd.to_datetime(
    df["Application Date"],
    errors="coerce",
    dayfirst=True
)
print("\nDate type after conversion:")
print(df["Application Date"].dtype)
print("\nMissing dates after conversion:")
print(df["Application Date"].isnull().sum())
display(df[["Application Date"]].head())
```

```python
df["Phone"] = (
    df["Phone"]
    .str.replace("-", "", regex=False)
    .str.replace(" ", "", regex=False)
)

print("Phone numbers after cleaning:")
display(df[["Phone"]].head(10))

print("\nMissing phone values:")
print(df["Phone"].isnull().sum())
```

I converted the Application Date column from text to a proper datetime format using `pd.to_datetime()` with `errors='coerce'` and `dayfirst=True`. Unreadable dates are converted to `NaT` so they can be identified and handled.

I removed dashes and spaces from phone numbers using `str.replace()`. Phone numbers remain stored as text because converting them to numbers could remove leading zeros.

I used `df.info()` to verify the data types after these changes.
